```python
# 레드 와인 데이터 확인 및 정리
red_df.head()
red_df.insert(0, columm = 'type', value = 'red')
red_df.head()
red_df.shape
```

```python
# 화이트 와인 데이터 확인 및 정리
white_df.head()
white_df.insert(0, columm = 'type', value = 'white')
white_df.head()
white_df.shape
```

```python
# 레드 와인과 화이트 와인 파일 합치기_concat()
wine = pd.concat([red_df, white_df])
wine.shape
wine.to_csv('Desktop/Git/파이썬프로그래밍/7장_data/wine.xlsx'
            , index = False)
```

```python
# 기본 정보 확인하기 info()
print(wine.info())
```

```python
# 함수를 사용해 기술 통계 구하기
wine.columns = wine.columns.str.replace(' ','_')
wine.head()
wine.describe()
```

```python
# 함수를 사용해 기술 통계 구하기
sorted(wine.quality.unique())
wine.quality.value_counts()
```

```python
# describe() 함수로 그룹 비교하기
wine.groupby('type')['quality'].describe()
wine.groupby('type')['quality'].mean()
wine.groupby('type')['quality'].std()
wine.groupby('type')['quality'].agg(['mean'],['std'])
```

```python
np.random.seed(0)

# 두 그룹 데이터 예시
group1 = np.random.normal(loc=50, scale=5, size=30)
group2 = np.random.normal(loc=55, scale=5, size=30)

# t-test
t_stat, p_value = stats.ttest_ind(group1, group2)

print("t-statistic:", t_stat)
print("p-value:", p_value)

#-------------------------
# 시각화 (박스플롯 + 히스토그램)
#-------------------------

plt.figure(figsize=(12, 5))

# 1) 박스플롯
plt.subplot(1, 2, 1)
plt.boxplot([group1, group2], labels=['Group 1', 'Group 2'])
plt.title("Boxplot of Two Groups")

# 2) 히스토그램
plt.subplot(1, 2, 2)
plt.hist(group1, bins=10, alpha=0.7, label='Group 1')
plt.hist(group2, bins=10, alpha=0.7, label='Group 2')
plt.legend()
plt.title("Histogram of Two Groups")

plt.show()
```
